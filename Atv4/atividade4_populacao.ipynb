```python
import pandas as pd
```

```python
df = pd.read_excel("CD2022_Populacao_2010_Compatibilizada.xlsx")
df.head()
```

```python
pop_uf = (
    df.groupby("UF")[["POP_2010_COMPATIBILIZADA", "POP_CENSO_2022"]]
    .sum()
    .reset_index()
)
pop_uf.head()
```

```python
# diferença entre a população de 2022 e a de 2010 (compatibilizada)
pop_uf["CRESCIMENTO_2010_2022"] = pop_uf["POP_CENSO_2022"] - pop_uf["POP_2010_COMPATIBILIZADA"]

pop_uf_ordenado = pop_uf.sort_values("CRESCIMENTO_2010_2022", ascending=False).reset_index(drop=True)
pop_uf_ordenado
```

```python
pop_uf_ordenado.to_csv("tabela_populacao_por_uf.csv", sep=";", index=False)
```

```python
pop_municipio = (
    df.groupby(["UF", "COD_MUNIC", "NOME_MUNICIPIO"])[["POP_2010_COMPATIBILIZADA", "POP_CENSO_2022"]]
    .sum()
    .reset_index()
)
pop_municipio.head()
```

```python
pop_municipio["CRESCIMENTO_2010_2022"] = pop_municipio["POP_CENSO_2022"] - pop_municipio["POP_2010_COMPATIBILIZADA"]

pop_municipio_ordenado = pop_municipio.sort_values("CRESCIMENTO_2010_2022", ascending=False).reset_index(drop=True)
pop_municipio_ordenado.head(10)
```

```python
pop_municipio_ordenado.to_csv("tabela_populacao_por_municipio.csv", sep=";", index=False)
```
