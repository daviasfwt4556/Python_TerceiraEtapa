```python
import pandas as pd
```

```python
df = pd.read_excel("Tabela_3_Areas_gerais_de_formacao_na_graduacao.xlsx")
df.head()
```

```python
curso_superior = df[["Região", "Total_Geral_Total", "Total_Geral_Homens", "Total_Geral_Mulheres"]].copy()
curso_superior.columns = ["Região", "Total", "Homens", "Mulheres"]
curso_superior
```

```python
tecnologia = df[["Região", "CTEM_Total", "CTEM_Homens", "CTEM_Mulheres"]].copy()
tecnologia.columns = ["Região", "Total", "Homens", "Mulheres"]
tecnologia
```

```python
educacao_saude = df[["Região", "ESSB_Total", "ESSB_Homens", "ESSB_Mulheres"]].copy()
educacao_saude.columns = ["Região", "Total", "Homens", "Mulheres"]
educacao_saude
```

```python
curso_superior.to_excel("curso_superior.xlsx", index=False)
tecnologia.to_excel("tecnologia.xlsx", index=False)
educacao_saude.to_excel("educacao_saude.xlsx", index=False)
```

```python
print("Curso superior (total geral):")
curso_superior.describe()
```

```python
print("Ciência, Tecnologia, Engenharias e Matemática:")
tecnologia.describe()
```

```python
print("Educação, Serviços pessoais, Saúde e Bem-estar:")
educacao_saude.describe()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

brasil = pd.DataFrame({
    "area": ["Graduação (total)", "Ciência e Tecnologia", "Educação e Saúde"],
    "homens": [
        curso_superior.loc[curso_superior["Região"] == "Brasil", "Homens"].item(),
        tecnologia.loc[tecnologia["Região"] == "Brasil", "Homens"].item(),
        educacao_saude.loc[educacao_saude["Região"] == "Brasil", "Homens"].item(),
    ],
    "mulheres": [
        curso_superior.loc[curso_superior["Região"] == "Brasil", "Mulheres"].item(),
        tecnologia.loc[tecnologia["Região"] == "Brasil", "Mulheres"].item(),
        educacao_saude.loc[educacao_saude["Região"] == "Brasil", "Mulheres"].item(),
    ],
})

brasil_long = brasil.melt(id_vars="area", var_name="sexo", value_name="pessoas")

fig, ax = plt.subplots(figsize=(9, 5))
sns.barplot(data=brasil_long, x="area", y="pessoas", hue="sexo", ax=ax)
ax.set_title("Brasil: pessoas com graduação, por área e sexo")
ax.set_xlabel("Área")
ax.set_ylabel("Pessoas")
plt.tight_layout()
plt.show()
```
