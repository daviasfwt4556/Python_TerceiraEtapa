```python
import matplotlib.pyplot as plt
import seaborn as sns

pop_estado = pd.read_csv("tabela_populacao_por_uf.csv", sep=";")

fig, ax = plt.subplots(figsize=(9, 8))
ordem = pop_estado.sort_values("POP_CENSO_2022", ascending=False)
sns.barplot(data=ordem, y="UF", x="POP_CENSO_2022", ax=ax, color="steelblue")
ax.set_title("População 2022 por UF")
ax.set_xlabel("Habitantes")
ax.set_ylabel("UF")
plt.tight_layout()
plt.show()
```
